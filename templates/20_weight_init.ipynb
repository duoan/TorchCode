> **Note:** `fan_in = weight.shape[1]` is correct for `nn.Linear` weights `(out_features, in_features)`. For a conv weight `(C_out, C_in, kH, kW)`, fan_in is `C_in * kH * kW`. `weight[0].numel()` gives the right value for both.

### 📐 Notation

| Symbol | Meaning |
|:---:|---|
| `C_out` | output channels (or output features for a linear layer) |
| `C_in` | input channels (or input features) |
| `kH` | kernel height |
| `kW` | kernel width |
| `fan_in` | number of inputs feeding each output unit |
