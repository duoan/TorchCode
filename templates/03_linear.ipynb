# 🟡 Medium: Simple Linear Layer

Implement a fully-connected linear layer: **y = xW^T + b**

### Signature
```python
class SimpleLinear(nn.Module):
    def __init__(self, in_features: int, out_features: int): ...
    def forward(self, x: torch.Tensor) -> torch.Tensor: ...
```

### Requirements
- Subclass `nn.Module` and call `super().__init__()` first
- `self.weight`: an `nn.Parameter` of shape `(out_features, in_features)`, init with `randn * (1/√in_features)`
- `self.bias`: an `nn.Parameter` of shape `(out_features,)`, init as zeros
- Both must be **registered parameters**, so `layer.parameters()` returns them and an optimizer can train them
- `forward(x)` computes `x @ W^T + b`; calling `layer(x)` must work too

> New to `nn.Module`? Prerequisite P8 covers parameters vs plain tensors and why registration matters.
- Do **NOT** use `torch.nn.Linear`
