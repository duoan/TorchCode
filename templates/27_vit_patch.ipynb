
### Algorithm
1. Reshape image into non-overlapping patches: `(B, C, H, W)` → `(B, N, C*P*P)`
2. Project each patch: `nn.Linear(C*P*P, embed_dim)`
3. `num_patches = (img_size // patch_size) ** 2`

### 📐 Notation

| Symbol | Meaning |
|:---:|---|
| `B` | batch size |
| `C` | input channels (3 for RGB) |
| `H` | image height |
| `W` | image width |
| `P` | patch size (`p` in the code) |
| `N` | number of patches, `(H / P) * (W / P)` |
| `embed_dim` | output dimension of each patch embedding |
